### Plot for PID control - static wall

```python
import matplotlib.pyplot as plt
import pandas as pd

f = pd.read_csv('HW1/Controls/csv/force_ctrl.csv')
i = pd.read_csv('HW1/Controls/csv/impedance_ctrl.csv')

fig, ax = plt.subplots()
f.plot(x='t', y='f', ax=ax, kind='line', label="Force Controller")
i.plot(x='t', y='f', ax=ax, kind='line', label="Impedance Controller")

ax.set_yticks(range(0, 196, 15))
ax.set_title("Force controller v.s. Impedance controller (static)")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Force (N)")
plt.show()
```

### Plot for PID control - oscillating wall

```python
import matplotlib.pyplot as plt
import pandas as pd

f = pd.read_csv('HW1/Controls/csv/force_ctrl_moving_wall.csv')
i = pd.read_csv('HW1/Controls/csv/impedance_ctrl_moving_wall.csv')

fig, ax = plt.subplots()
f.plot(x='t', y='f', ax=ax, kind='line', label="Force Controller")
i.plot(x='t', y='f', ax=ax, kind='line', label="Impedance Controller")

ax.set_yticks(range(0, 196, 15))
ax.set_title("Force controller v.s. Impedance controller (oscillating)")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Force (N)")

plt.show()
```
